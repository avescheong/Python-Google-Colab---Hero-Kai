## 🧙 The Final Spell
Run the cell below to watch Kai cast all Eleven Ancient Spells in one legendary combo.

```python
# ================================
#   THE FALL OF THE BUG KING
# ================================

# Spell 1: VOICE
print("Kai raises his staff and speaks: \"Bug King, your reign of silence ends today!\"")

# Spell 2: WEAVE
print("Eleven", "spells,", "one", "apprentice,", "one", "final", "stand.")

# Spell 3: FUSE
print("Py" + "thon" + " magic surges through Kai's hands!")

# Spell 4: LISTEN
final_word = input("The Great Wizard asks: What word will you cast to end this battle? ")
print("Kai channels the word:", final_word)

# Spell 5 & 6: VAULT + MULTI-VAULT
hp, mana, gold = 100, 100, 100
print("Kai's power is fully restored! HP:", hp, "| Mana:", mana, "| Gold:", gold)

# Spell 7: WHISPER
# The Great Wizard whispers one last secret: "Believe in yourself, Kai."
print("A silent whisper of courage fills the air.")

# Spell 8: SHAPESHIFT
curse_type = type(hp)
print("The Bug King's curse is nothing but a fragile", curse_type, "- and Kai knows exactly how to break it.")

# Spell 9: FORGE
final_power = (hp + mana + gold) * 3
print("Kai forges all his strength together! Final Power:", final_power)

# Spell 10: SIGHT
print("Is Kai's power greater than the Bug King's curse (300)?", final_power > 300)

# Spell 11: FATE
print()
if final_power > 300:
    print("🎉 THE CURSE IS BROKEN! The Bug King shatters into a thousand harmless bugs!")
    print("Color floods back into the Kingdom of Pythonia. Birds sing. Children laugh.")
    victory = True
elif final_power == 300:
    print("⚡ A perfect tie! Kai and the Bug King vanish in a flash of light...")
    victory = None
else:
    print("💫 Not quite enough power yet - but Kai refuses to give up!")
    victory = False

print()
print("========================")
print("   BATTLE REPORT")
print("========================")
print("Final word cast:", final_word)
print("Final Power:", final_power)
print("Victory?", victory)
print("========================")
```
