```python
import time
import random

# ==========================================
# ⚠️ AI FOR GOOD ZONE: ADJUST PARAMETERS HERE ⚠️
# ==========================================

# The Global Hazard
crisis_name = "The Great Wildfire"
crisis_hp = 500  # The fire's total size. Try 2000 for a massive challenge!
crisis_damage = 30 # How much damage the fire does to the drones each turn

# The Team Variables
team_integrity = 100  # Total team health/battery life
tool = "Environmental Assessment Drones"

# This list represents the whole class!
# Each number is a different student's individual "Power" variable.
# Add or change numbers to represent more students.
student_powers = [12, 8, 15]

# ==========================================
# DO NOT TOUCH THE CODE BELOW UNLESS YOU ARE A MASTER CODER
# ==========================================

print("-----------------------------------")
print(f"🌍 GLOBAL ALERT: {crisis_name} DETECTED! ")
print("-----------------------------------")
time.sleep(1)
print(f"Deploying {len(student_powers)} AI Agents piloting {tool}...")
time.sleep(1.5)

while team_integrity > 0 and crisis_hp > 0:
    print("\n[ TEAM ACTION TURN ] >>>")
    time.sleep(1)

    # ⚙️ THE LOOP: Calculating Total Class Power
    total_class_impact = 0
    for power in student_powers:
        # Each drone calculates an impact between 1 and their specific max power
        drone_impact = random.randint(1, power)
        total_class_impact = total_class_impact + drone_impact

    # Subtract the combined class impact from the fire
    crisis_hp = crisis_hp - total_class_impact

    print(f"💦 The drone swarm works together and extinguishes {total_class_impact} fire points!")

    if crisis_hp <= 0:
        crisis_hp = 0
        print(f"📉 {crisis_name} size is now: {crisis_hp}")
        print("\n🎉 O P E R A T I O N  S U C C E S S ! 🎉")
        print("The ecosystem has been saved by the combined power of the class!")
        break

    print(f"📉 {crisis_name} size is now: {crisis_hp}")
    time.sleep(1.5)

    print("\n[ HAZARD EXPANSION ] >>>")
    time.sleep(1)

    # Fire Attack Math
    print(f"🔥 {crisis_name} spreads! The extreme heat damages the drone swarm by {crisis_damage}!")
    team_integrity = team_integrity - crisis_damage

    if team_integrity <= 0:
        team_integrity = 0
        print(f"🛡️ Team Integrity is now: {team_integrity}")
        print("\n💀 S Y S T E M  F A I L U R E 💀")
        print("The fire overwhelmed the drones. Tweak your team variables and try again...")
        break

    print(f"🛡️ Team Integrity is now: {team_integrity}")
    time.sleep(1.5)
```
